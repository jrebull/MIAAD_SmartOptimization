<h2 style="color:#0F3460; font-family:Georgia, serif; border-left: 4px solid #E94560; padding-left: 12px; margin-top: 30px;">1. Definición Matemática del Modelo</h2>

<p style="font-size:15px; color:#333; line-height:1.7;">
Buscamos maximizar el <b>Margen Bruto Total</b> ($Z$). Para ello, calculamos la diferencia entre el ingreso por venta de maíz y el costo del fertilizante aplicado en cada zona.
</p>

<p style="font-size:15px; color:#333;">Nuestra función objetivo es:</p>

$$ Z = \sum_{i=1}^{10} \left[ (P_m \cdot Y_i(N_i) - C_n \cdot N_i) \cdot Area_i \right] $$

<p style="font-size:15px; color:#333;">Donde la respuesta del cultivo ($Y_i$) sigue una curva cuadrática de rendimientos decrecientes:</p>

$$ Y_i(N_i) = a_i N_i^2 + b_i N_i + c_i $$

<div style="background:#F0F3F7; border-radius:8px; padding:15px 20px; margin:15px 0;">
<p style="margin:0 0 8px 0; font-weight:bold; color:#0F3460;">Parámetros económicos</p>
<ul style="margin:0; color:#333;">
    <li>$P_m$: Precio del maíz = <b>200 USD/ton</b></li>
    <li>$C_n$: Costo del nitrógeno = <b>1.2 USD/kg</b></li>
    <li>$Area_i$: Superficie por zona = <b>10 ha</b></li>
</ul>
</div>

<h2 style="color:#003CA6;">1. Modelado de Respuesta del Cultivo al Nitrógeno</h2>
<p>La relación entre la dosis de fertilizante nitrogenado y el rendimiento del maíz no es lineal. Según la teoría agronómica, esta relación sigue la <b>Ley de los Rendimientos Decrecientes</b>. Diversos estudios han demostrado que los modelos polinomiales cuadráticos son efectivos para describir esta respuesta, donde el rendimiento aumenta hasta un punto óptimo y luego decrece debido a la toxicidad o salinidad (Cerrato & Blackmer, 1990).</p>
<p>Matemáticamente, esto se expresa como:</p>

$$ Y(N) = aN^2 + bN + c $$

<div style="background-color:#EBF0F7; padding: 15px; border-left: 5px solid #003CA6;">
    <p><b>Nota Clave:</b> El coeficiente cuadrático ($a$) debe ser negativo ($a < 0$) para reflejar la convexidad de la curva y el efecto de saturación del nutriente.</p>
</div>

<h2 style="color:#003CA6;">2. Algoritmos Genéticos de Codificación Real (RCGA)</h2>
<p>Los Algoritmos Genéticos (AG) tradicionales utilizan cadenas binarias (0s y 1s). Sin embargo, para problemas de ingeniería con variables continuas (como la cantidad exacta de kg/ha), la codificación binaria presenta dificultades de precisión y el "fenómeno de Hamming".</p>
<p>Los <b>Algoritmos Genéticos de Codificación Real (RCGA)</b> utilizan vectores de números flotantes para representar los cromosomas. Según Deb (2001), esta representación permite explotar la gradiente local de la función objetivo de manera más eficiente y evita la necesidad de decodificación, mejorando la velocidad de convergencia en espacios de búsqueda continuos.</p>

<h2 style="color:#003CA6;">3. Manejo de Restricciones por Penalización</h2>
<p>En problemas de optimización restringida, es común que el óptimo global se encuentre en los límites de la región factible. Simplemente descartar soluciones inviables es ineficiente.</p>
<p>Coello (2002) establece que el método de <b>Funciones de Penalización Exterior</b> es una de las estrategias más robustas. Este método transforma el problema restringido en uno no restringido mediante la adición de un término de castigo a la función objetivo, proporcional a la magnitud de la violación:</p>

$$ F_{eval}(x) = f(x) - \sum_{j=1}^{m} K_j \cdot [\max(0, g_j(x))]^2 $$

<p>El uso de una penalización cuadrática asegura que pequeñas violaciones sean toleradas en etapas tempranas de la evolución, favoreciendo la diversidad genética, mientras que las grandes violaciones son severamente castigadas.</p>

## 1. Configuración Inicial y Carga de Datos

#### Sobre `deepcopy`

La función `deepcopy` del módulo estándar `copy` de Python permite crear una **copia completamente independiente** de un objeto, incluyendo todos los objetos anidados dentro de él. A diferencia de una copia superficial (`copy`), donde los sub-objetos siguen apuntando a la misma referencia en memoria, `deepcopy` genera nuevas instancias en todos los niveles de profundidad.

En el contexto de nuestro Algoritmo Genético, esto es fundamental para **preservar al mejor individuo (élite)** entre generaciones sin que las operaciones de cruza o mutación sobre la población modifiquen accidentalmente su contenido. Sin `deepcopy`, copiar un cromosoma solo crearía una referencia al mismo array, y cualquier cambio en la población alteraría también al élite guardado.

> **Referencia:** Python Software Foundation. *copy — Shallow and deep copy operations.* Documentación oficial de Python 3.12. Disponible en: [https://docs.python.org/3/library/copy.html](https://docs.python.org/3/library/copy.html)

## 2. Funciones del Modelo Matemático

```python
def rendimiento(N, a, b, c):
    """
    Calcula el rendimiento Y_i (ton/ha) para cada zona.
    Y_i(N_i) = a_i * N_i^2 + b_i * N_i + c_i
    """
    return a * N**2 + b * N + c


def margen_bruto(N):
    """
    Calcula el Margen Bruto Total Z(X).
    Z = Σ (P_m * Y_i(N_i) - C_n * N_i) * Area_i
    """
    Y = rendimiento(N, a_coef, b_coef, c_coef)
    ingresos = P_MAIZ * Y * areas       # USD por zona
    costos   = C_NITRO * N * areas       # USD por zona
    return np.sum(ingresos - costos)


def costo_total(N):
    """Costo total de fertilizante en USD."""
    return np.sum(C_NITRO * N * areas)


def produccion_total(N):
    """Producción total en toneladas."""
    Y = rendimiento(N, a_coef, b_coef, c_coef)
    return np.sum(Y * areas)


def fitness_penalizado(N, K_pres=30.0, K_prod=25.0):
    """
    Fitness con penalización exterior cuadrática.

    F_eval = Z(X) - K_pres * max(0, C_T - C_max)^2
                   - K_prod * max(0, Prod_min - Prod_total)^2

    Parámetros:
        N      : array de dosis de nitrógeno (10 valores)
        K_pres : factor de penalización por presupuesto
        K_prod : factor de penalización por producción mínima
    """
    Z = margen_bruto(N)

    # Penalización por presupuesto
    C_T = costo_total(N)
    viol_pres = max(0, C_T - PRES_MAX)
    pen_pres = K_pres * viol_pres**2

    # Penalización por producción mínima
    prod = produccion_total(N)
    viol_prod = max(0, PROD_MIN - prod)
    pen_prod = K_prod * viol_prod**2

    return Z - pen_pres - pen_prod


print("Funciones del modelo matemático definidas.")
```

## 3. Validación: Ejemplo del Documento

Reproducimos el ejemplo de la guía con el cromosoma `X = [180, 180, 120, 120, 120, 80, 80, 80, 40, 40]` para verificar que nuestras funciones calculan correctamente.

```python
# ============================================================
# VALIDACIÓN CON EL EJEMPLO DE LA GUÍA
# ============================================================

X_ejemplo = np.array([180, 180, 120, 120, 120, 80, 80, 80, 40, 40], dtype=float)

print("VALIDACIÓN: Cromosoma de Ejemplo")
print("=" * 70)
print(f"  X = {X_ejemplo.astype(int).tolist()}")
print()

# Rendimiento por zona
Y_vals = rendimiento(X_ejemplo, a_coef, b_coef, c_coef)

print(" Paso 1 — Rendimiento por zona (ton/ha):")
print("-" * 70)
for i in range(N_ZONAS):
    tipo = datos_zonas['Tipo_Suelo'].iloc[i]
    print(f"  Zona {i+1:2d} ({tipo:17s}): N={X_ejemplo[i]:5.0f} kg/ha → "
          f"Y = {a_coef[i]:.5f}·{X_ejemplo[i]:.0f}² + {b_coef[i]:.3f}·{X_ejemplo[i]:.0f} + {c_coef[i]:.1f} "
          f"= {Y_vals[i]:.4f} ton/ha")

print(f"\n  ✔ Zona 1 (doc): Y₁ = 24.928 ton/ha → Calculado: {Y_vals[0]:.3f} ton/ha")
print(f"  ✔ Zona 9 (doc): Y₉ =  3.620 ton/ha → Calculado: {Y_vals[8]:.3f} ton/ha")

# Evaluación económica
print("\n Paso 2 — Evaluación Económica:")
print("-" * 70)
ingresos_zona = P_MAIZ * Y_vals * areas
costos_zona   = C_NITRO * X_ejemplo * areas
margen_zona   = ingresos_zona - costos_zona

for i in range(N_ZONAS):
    print(f"  Zona {i+1:2d}: Ingreso=${ingresos_zona[i]:>10,.2f}  "
          f"Costo=${costos_zona[i]:>8,.2f}  "
          f"Margen=${margen_zona[i]:>10,.2f}")

Z_total = margen_bruto(X_ejemplo)
C_total = costo_total(X_ejemplo)
P_total = produccion_total(X_ejemplo)

print(f"\n  RESUMEN:")
print(f"  ├─ Margen Bruto Total (Z):   ${Z_total:>12,.2f} USD")
print(f"  ├─ Costo Total Fertilizante: ${C_total:>12,.2f} USD")
print(f"  ├─ Producción Total:          {P_total:>10,.2f} toneladas")
print(f"  │")

# Verificación de restricciones
print(f"  Paso 3 — Restricciones:")
ok_pres = "✅ CUMPLE" if C_total <= PRES_MAX else f"❌ EXCEDE por ${C_total - PRES_MAX:,.2f}"
ok_prod = "✅ CUMPLE" if P_total >= PROD_MIN else f"❌ FALTA {PROD_MIN - P_total:,.2f} ton"
print(f"  ├─ Presupuesto (≤ ${PRES_MAX:,.0f}):   {ok_pres}")
print(f"  └─ Producción  (≥ {PROD_MIN:.0f} ton):  {ok_prod}")

# Fitness penalizado
F_eval = fitness_penalizado(X_ejemplo)
print(f"\n  Fitness Penalizado: ${F_eval:,.2f} USD")
```

## 4. Visualización: Curvas de Respuesta del Cultivo

## 5. Implementación del Algoritmo Genético

### 5.1 Operadores Genéticos

```python
# ============================================================
# OPERADORES DEL ALGORITMO GENÉTICO
# ============================================================

def inicializar_poblacion(n_pop, n_genes, lb, ub):
    """
    Genera población inicial aleatoria dentro de los límites.
    Cada individuo es un vector de dosis N_i ∈ [lb, ub].
    """
    return np.random.uniform(lb, ub, size=(n_pop, n_genes))


def seleccion_torneo(poblacion, fitness_vals, k=3):
    """
    Selección por torneo de tamaño k.
    Selecciona k individuos al azar y retorna el de mayor fitness.
    """
    indices = np.random.randint(0, len(poblacion), size=k)
    mejor_idx = indices[np.argmax(fitness_vals[indices])]
    return poblacion[mejor_idx].copy()


def cruza_sbx(padre1, padre2, eta_c=20, prob_cruza=0.9, lb=N_MIN, ub=N_MAX):
    """
    Simulated Binary Crossover (SBX).

    Genera dos hijos a partir de dos padres usando el operador SBX
    con índice de distribución η_c.

    Parámetros:
        padre1, padre2 : arrays de genes
        eta_c          : índice de distribución (mayor = hijos más cercanos a padres)
        prob_cruza     : probabilidad de aplicar cruza
        lb, ub         : límites inferior y superior
    """
    hijo1 = padre1.copy()
    hijo2 = padre2.copy()

    if np.random.random() > prob_cruza:
        return hijo1, hijo2

    for i in range(len(padre1)):
        if np.random.random() > 0.5:
            continue

        if abs(padre1[i] - padre2[i]) < 1e-14:
            continue

        # Cálculo del factor beta
        if padre1[i] < padre2[i]:
            y1, y2 = padre1[i], padre2[i]
        else:
            y1, y2 = padre2[i], padre1[i]

        # Beta para respetar límites
        beta_l = 1.0 + (2.0 * (y1 - lb) / (y2 - y1))
        beta_u = 1.0 + (2.0 * (ub - y2) / (y2 - y1))

        def calc_betaq(beta_val):
            alpha = 2.0 - beta_val ** (-(eta_c + 1.0))
            u = np.random.random()
            if u <= 1.0 / alpha:
                return (u * alpha) ** (1.0 / (eta_c + 1.0))
            else:
                return (1.0 / (2.0 - u * alpha)) ** (1.0 / (eta_c + 1.0))

        betaq = min(calc_betaq(beta_l), calc_betaq(beta_u))

        # Generar hijos
        hijo1[i] = 0.5 * ((y1 + y2) - betaq * (y2 - y1))
        hijo2[i] = 0.5 * ((y1 + y2) + betaq * (y2 - y1))

        # Reparar límites
        hijo1[i] = np.clip(hijo1[i], lb, ub)
        hijo2[i] = np.clip(hijo2[i], lb, ub)

    return hijo1, hijo2


def mutacion_polinomial(individuo, prob_mut=0.1, eta_m=20, lb=N_MIN, ub=N_MAX):
    """
    Mutación Polinomial.

    Aplica perturbación polinomial a cada gen con probabilidad p_m.

    Parámetros:
        individuo : array de genes
        prob_mut  : probabilidad de mutar cada gen
        eta_m     : índice de distribución de mutación
        lb, ub    : límites inferior y superior
    """
    mutado = individuo.copy()

    for i in range(len(mutado)):
        if np.random.random() >= prob_mut:
            continue

        y = mutado[i]
        delta1 = (y - lb) / (ub - lb)
        delta2 = (ub - y) / (ub - lb)

        u = np.random.random()
        if u < 0.5:
            xy = 1.0 - delta1
            val = 2.0 * u + (1.0 - 2.0 * u) * (xy ** (eta_m + 1.0))
            deltaq = val ** (1.0 / (eta_m + 1.0)) - 1.0
        else:
            xy = 1.0 - delta2
            val = 2.0 * (1.0 - u) + 2.0 * (u - 0.5) * (xy ** (eta_m + 1.0))
            deltaq = 1.0 - val ** (1.0 / (eta_m + 1.0))

        mutado[i] = y + deltaq * (ub - lb)
        mutado[i] = np.clip(mutado[i], lb, ub)

    return mutado


print("✅ Operadores genéticos definidos:")
print("   • Inicialización aleatoria uniforme")
print("   • Selección por torneo (k=3)")
print("   • Cruza SBX (η_c=20)")
print("   • Mutación polinomial (η_m=20, p_m=0.1)")
```

### 5.2 Hiperparámetros del Algoritmo Genético

```python
# ============================================================
# HIPERPARÁMETROS
# ============================================================

N_POP         = 200      # Tamaño de población
N_GEN         = 500      # Número de generaciones
PROB_CRUZA    = 0.9      # Probabilidad de cruza
PROB_MUT      = 0.1      # Probabilidad de mutación por gen
ETA_C         = 20       # Índice de distribución SBX
ETA_M         = 20       # Índice de distribución mutación
TORNEO_K      = 3        # Tamaño de torneo
K_PRESUPUESTO = 30.0     # Factor de penalización presupuesto
K_PRODUCCION  = 25.0     # Factor de penalización producción

# Semilla para reproducibilidad
SEED = 42
np.random.seed(SEED)

print("⚙️  Hiperparámetros del AG")
print("=" * 45)
print(f"  Población:        {N_POP}")
print(f"  Generaciones:     {N_GEN}")
print(f"  P(cruza):         {PROB_CRUZA}")
print(f"  P(mutación):      {PROB_MUT}")
print(f"  η_c (SBX):        {ETA_C}")
print(f"  η_m (mutación):   {ETA_M}")
print(f"  Torneo k:         {TORNEO_K}")
print(f"  K_presupuesto:    {K_PRESUPUESTO}")
print(f"  K_producción:     {K_PRODUCCION}")
print(f"  Semilla:          {SEED}")
```

### 5.3 Loop Evolutivo Principal

```python
# ============================================================
# ALGORITMO GENÉTICO — LOOP EVOLUTIVO
# ============================================================

np.random.seed(SEED)

# Inicializar población
poblacion = inicializar_poblacion(N_POP, N_ZONAS, N_MIN, N_MAX)

# Historial para gráficas de convergencia
hist_mejor_fitness  = []
hist_prom_fitness   = []
hist_peor_fitness   = []
hist_mejor_individuo = []
hist_diversidad     = []

# Mejor global
mejor_global = None
mejor_global_fitness = -np.inf

print("🚀 Iniciando Algoritmo Genético...")
print("=" * 70)

for gen in range(N_GEN):

    # --- Evaluar fitness de toda la población ---
    fitness_vals = np.array([
        fitness_penalizado(ind, K_PRESUPUESTO, K_PRODUCCION)
        for ind in poblacion
    ])

    # --- Registrar estadísticas ---
    idx_mejor = np.argmax(fitness_vals)
    mejor_fit = fitness_vals[idx_mejor]
    prom_fit  = np.mean(fitness_vals)
    peor_fit  = np.min(fitness_vals)

    hist_mejor_fitness.append(mejor_fit)
    hist_prom_fitness.append(prom_fit)
    hist_peor_fitness.append(peor_fit)
    hist_mejor_individuo.append(poblacion[idx_mejor].copy())
    hist_diversidad.append(np.mean(np.std(poblacion, axis=0)))

    # Actualizar mejor global
    if mejor_fit > mejor_global_fitness:
        mejor_global_fitness = mejor_fit
        mejor_global = poblacion[idx_mejor].copy()

    # --- Imprimir progreso ---
    if gen % 50 == 0 or gen == N_GEN - 1:
        ct = costo_total(poblacion[idx_mejor])
        pt = produccion_total(poblacion[idx_mejor])
        print(f"  Gen {gen:4d}/{N_GEN} │ Mejor: ${mejor_fit:>12,.2f} │ "
              f"Prom: ${prom_fit:>12,.2f} │ Costo: ${ct:>8,.0f} │ Prod: {pt:>7,.1f} ton")

    # --- Elitismo: preservar al mejor ---
    elite = poblacion[idx_mejor].copy()

    # --- Crear nueva población ---
    nueva_poblacion = [elite]  # El élite pasa directamente

    while len(nueva_poblacion) < N_POP:
        # Selección
        padre1 = seleccion_torneo(poblacion, fitness_vals, TORNEO_K)
        padre2 = seleccion_torneo(poblacion, fitness_vals, TORNEO_K)

        # Cruza SBX
        hijo1, hijo2 = cruza_sbx(padre1, padre2, ETA_C, PROB_CRUZA, N_MIN, N_MAX)

        # Mutación polinomial
        hijo1 = mutacion_polinomial(hijo1, PROB_MUT, ETA_M, N_MIN, N_MAX)
        hijo2 = mutacion_polinomial(hijo2, PROB_MUT, ETA_M, N_MIN, N_MAX)

        nueva_poblacion.append(hijo1)
        if len(nueva_poblacion) < N_POP:
            nueva_poblacion.append(hijo2)

    poblacion = np.array(nueva_poblacion[:N_POP])

print("\n" + "=" * 70)
print("✅ Algoritmo Genético finalizado.")
print(f"🏆 Mejor fitness global: ${mejor_global_fitness:,.2f} USD")
```

## 7. Análisis de Sensibilidad: Factor de Penalización

```python
# ============================================================
# ANÁLISIS DE SENSIBILIDAD SOBRE K_PRESUPUESTO
# ============================================================

K_values = [5, 10, 20, 30, 50, 100]
resultados_K = []

print("🔬 Análisis de Sensibilidad: Factor K de Penalización")
print("=" * 80)

for K_test in K_values:
    np.random.seed(SEED)
    pop_test = inicializar_poblacion(N_POP, N_ZONAS, N_MIN, N_MAX)
    mejor_test = None
    mejor_fit_test = -np.inf

    for gen in range(300):  # 300 generaciones rápidas
        fit_vals = np.array([fitness_penalizado(ind, K_test, K_PRODUCCION) for ind in pop_test])
        idx_m = np.argmax(fit_vals)

        if fit_vals[idx_m] > mejor_fit_test:
            mejor_fit_test = fit_vals[idx_m]
            mejor_test = pop_test[idx_m].copy()

        elite_t = pop_test[idx_m].copy()
        nueva_pop = [elite_t]

        while len(nueva_pop) < N_POP:
            p1 = seleccion_torneo(pop_test, fit_vals, TORNEO_K)
            p2 = seleccion_torneo(pop_test, fit_vals, TORNEO_K)
            h1, h2 = cruza_sbx(p1, p2, ETA_C, PROB_CRUZA)
            h1 = mutacion_polinomial(h1, PROB_MUT, ETA_M)
            h2 = mutacion_polinomial(h2, PROB_MUT, ETA_M)
            nueva_pop.append(h1)
            if len(nueva_pop) < N_POP:
                nueva_pop.append(h2)
        pop_test = np.array(nueva_pop[:N_POP])

    ct = costo_total(mejor_test)
    pt = produccion_total(mejor_test)
    mb = margen_bruto(mejor_test)
    resultados_K.append({
        'K': K_test,
        'Margen Bruto': mb,
        'Costo Total': ct,
        'Producción': pt,
        'Factible': '✅' if (ct <= PRES_MAX and pt >= PROD_MIN) else '❌'
    })
    print(f"  K={K_test:>4d} │ Margen: ${mb:>10,.2f} │ Costo: ${ct:>8,.0f} │ "
          f"Prod: {pt:>7,.1f} ton │ {'✅ Factible' if ct <= PRES_MAX and pt >= PROD_MIN else '❌ Infactible'}")

df_sensibilidad = pd.DataFrame(resultados_K)
print("\n📋 Tabla de Sensibilidad:")
df_sensibilidad
```
